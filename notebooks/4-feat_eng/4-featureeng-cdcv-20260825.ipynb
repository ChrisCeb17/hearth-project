### 4.2 Valores faltantes

```python
nulos = heart_features.isna().sum()
porcentaje_nulos = (nulos / len(heart_features) * 100).round(2)
pd.DataFrame({"nulos": nulos, "% nulos": porcentaje_nulos}).sort_values(
    "% nulos", ascending=False
)
```

_Nota: los valores faltantes **no se imputan aquí todavía** — la imputación se hace dentro
del pipeline de scikit-learn (más abajo), para evitar fuga de información (data leakage) entre
train y test._

### 4.3 Registros duplicados

```python
duplicate_rows = heart_features.duplicated().sum()
print(
    "Filas duplicadas:",
    duplicate_rows,
    f"({duplicate_rows / len(heart_features) * 100:.2f}%)",
)

heart_features = heart_features.drop_duplicates()
print("Shape después de eliminar duplicados:", heart_features.shape)
```

### 4.4 Valores atípicos (outliers)

Revisión rápida con IQR sobre las variables numéricas (identificadas más abajo). Por ahora solo se reportan; **decidir si se corrigen, se eliminan o se dejan** según el problema (en un dataset clínico, valores extremos pueden ser reales y clínicamente relevantes, no ruido).

```python
numeric_cols_check = list(heart_features.select_dtypes(include=["number"]).columns)
if TARGET in numeric_cols_check:
    numeric_cols_check.remove(TARGET)

for col in numeric_cols_check:
    q1, q3 = heart_features[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lim_inf, lim_sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_outliers = (
        (heart_features[col] < lim_inf) | (heart_features[col] > lim_sup)
    ).sum()
    print(
        f"{col}: {n_outliers} outliers ({n_outliers / len(heart_features) * 100:.2f}%) "
        f"fuera de [{lim_inf:.2f}, {lim_sup:.2f}]"
    )
```

## ✂️ 5. Train / Test split

Se hace el split **antes** de ajustar cualquier transformador, para que el pipeline aprenda (imputación, escalado, encoding) solo con datos de entrenamiento y no filtre información del test.

```python
X_features = heart_features.drop(columns=[TARGET])
y_target = heart_features[TARGET]

x_train, x_test, y_train, y_test = train_test_split(
    X_features, y_target, test_size=0.2, stratify=y_target, random_state=42
)

print("Train:", x_train.shape, y_train.shape)
print("Test:", x_test.shape, y_test.shape)
```

## 🔧 7. Pipelines de scikit-learn

```python
numeric_pipe = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]
)

categorical_pipe = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ]
)

categorical_ord_pipe = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("ordinal", OrdinalEncoder()),
    ]
)

transformers = [("numeric", numeric_pipe, cols_numeric)]

if cols_categoric:
    transformers.append(("categoric", categorical_pipe, cols_categoric))

if cols_categoric_ord:
    transformers.append(("categoric_ordinal", categorical_ord_pipe, cols_categoric_ord))

preprocessor = ColumnTransformer(transformers=transformers)
preprocessor
```

### 7.1 Ajustar y transformar

```python
print(x_train.dtypes)
print()
print("cols_numeric:", cols_numeric)
print("cols_categoric:", cols_categoric)
print("cols_categoric_ord:", cols_categoric_ord)
```

```python
preprocessor.fit(x_train)

feature_names = preprocessor.get_feature_names_out()

x_train_transformed = preprocessor.transform(x_train)
x_train_transformed = pd.DataFrame(
    x_train_transformed, columns=feature_names, index=x_train.index
)

x_train_transformed.info()
```
