## 💾 Load data

```python
###null values
df_heart = df_heart.replace("?", np.nan)

df_heart.info()
```

```python
# Assuming your DataFrame is named 'df'
# df = pd.read_csv('your_file.csv')

# Identify numeric and categorical columns
numeric_cols = ["age", "rest_bp", "chol", "max_hr", "old_peak"]
categorical_cols = ["sex", "chest_pain", "disease"]

# Convert numeric columns
df_heart[numeric_cols] = df_heart[numeric_cols].apply(pd.to_numeric, errors="coerce")

# Convert categorical columns
df_heart[categorical_cols] = df_heart[categorical_cols].astype("category")

# Print the updated data types
print(df_heart.dtypes)
```

```python
df_heart.info()
```

```python
# ============================================================
# CLASIFICACIÓN DE VARIABLES DEL DATASET CORAZON
# ============================================================

clasificacion = pd.DataFrame(
    {
        "columna": [
            "age",
            "sex",
            "chest_pain",
            "rest_bp",
            "chol",
            "fbs",
            "rest_ecg",
            "max_hr",
            "exang",
            "old_peak",
            "slope",
            "ca",
            "thal",
            "disease",
        ],
        # ⚠️ Cambiado de df.dtypes a df_heart.dtypes
        "tipo_original": df_heart.dtypes.values.astype(str),
        "tipo_correcto": [
            "int8",
            "category nominal",
            "category nominal",
            "int16",
            "int16",
            "bool",
            "category nominal",
            "int16",
            "bool",
            "float32",
            "category ordinal",
            "Int8 nullable",
            "category nominal",
            "bool",
        ],
        "descripcion": [
            "Edad del paciente",
            "Sexo: Male / Female",
            "Tipo de dolor: typical, nontypical, nonanginal, asymptomatic",
            "Presión arterial en reposo (mm Hg)",
            "Colesterol sérico (mg/dl)",
            "Azúcar en ayunas > 120 mg/dl",
            "Resultados del ECG en reposo",
            "Frecuencia cardíaca máxima alcanzada",
            "Angina inducida por ejercicio",
            "Depresión del ST inducida por ejercicio",
            "Pendiente del ST: 1=ascendente, 2=plana, 3=descendente",
            "Número de vasos coloreados por fluoroscopía (0-3)",
            "Prueba de estrés: normal, fixed, reversable",
            "Diagnóstico de enfermedad cardíaca (target)",
        ],
    }
)

# ============================================================
# CONVERSIÓN DE TIPOS DE DATOS SEGÚN CLASIFICACIÓN
# ============================================================

for _, row in clasificacion.iterrows():
    col = row["columna"]
    tipo = row["tipo_correcto"]

    try:
        if tipo == "bool":
            # Convertir a bool: 1 → True, 0 → False
            df_heart[col] = pd.to_numeric(df_heart[col], errors="coerce").astype("boolean")

        elif tipo.startswith("category"):
            # Convertir a categoría (nominal u ordinal)
            df_heart[col] = df_heart[col].astype("category")

        elif tipo == "Int8 nullable":
            # Entero de 8 bits que admite NaN (mayúscula = nullable)
            df_heart[col] = pd.to_numeric(df_heart[col], errors="coerce").astype("Int8")

        elif tipo.startswith("int"):
            # Primero convertir a float para manejar NaN, luego a Int nullable
            tipo_nullable = tipo.replace("int", "Int")  # int8 → Int8, int16 → Int16
            df_heart[col] = pd.to_numeric(df_heart[col], errors="coerce").astype(tipo_nullable)

        elif tipo.startswith("float"):
            # Convertir a flotante (admite NaN por defecto)
            df_heart[col] = pd.to_numeric(df_heart[col], errors="coerce").astype(tipo)

        print(f"✅ {col:12} → {tipo}")

    except Exception as e:
        print(f"❌ {col:12} → Error: {e}")

# ============================================================
# VERIFICAR TIPOS RESULTANTES
# ============================================================

print("\n--- Tipos de datos después de la conversión ---")
print(df_heart.dtypes)

print("\n--- Clasificación completa ---")
clasificacion["tipo_final"] = df_heart.dtypes.values.astype(str)
clasificacion
```

```python
# ============================================================
# ELIMINAR FILAS CON VALORES INVÁLIDOS EN COLUMNAS CATEGÓRICAS
# ============================================================
# Estado actual de los tipos de datos (resultado de clasificacion):
# sex        → category     (ya convertida)
# chest_pain → category     (ya convertida)
# rest_ecg   → str          (pendiente conversión)
# slope      → str          (pendiente conversión)
# thal       → str          (pendiente conversión)
# exang      → str          (pendiente conversión)
# fbs        → float64      (ya numérica)
# disease    → category     (ya convertida)
# ca         → str          (pendiente conversión)
# ============================================================

print("=" * 60)
print("LIMPIEZA DE VALORES INVÁLIDOS EN COLUMNAS CATEGÓRICAS")
print("=" * 60)
print(f"📊 Filas antes de la limpieza: {len(df_heart)}\n")

# ------------------------------------------------------------
# 1. COLUMNAS YA TIPO CATEGORY (sex, chest_pain, disease)
#    Solo verificar que no tengan categorías inválidas
# ------------------------------------------------------------

valores_validos_category = {
    "sex": ["Male", "Female"],
    "chest_pain": ["typical", "nontypical", "nonanginal", "asymptomatic"],
    "disease": [0, 1, True, False],  # ya es boolean/category
}

print("--- Columnas tipo category ---")
for col, validos in valores_validos_category.items():
    # Convertir a str para comparar de forma segura
    mask = ~df_heart[col].astype(str).isin([str(v) for v in validos]) & df_heart[col].notna()
    n = mask.sum()
    print(f"  {'✅' if n == 0 else '❌'} {col:12}: {n:4} filas inválidas eliminadas")
    if n > 0:
        print(f"     Inválidos encontrados: {df_heart.loc[mask, col].unique().tolist()}")
    df_heart = df_heart[~mask].reset_index(drop=True)

# ------------------------------------------------------------
# 2. COLUMNAS TIPO STR (rest_ecg, thal)
#    Validar valores y luego convertir a category
# ------------------------------------------------------------

valores_validos_str = {
    "rest_ecg": [
        "normal",
        "left ventricular hypertrophy",
        "ST-T wave abnormality",
    ],
    "thal": ["normal", "fixed", "reversable"],
}

print("\n--- Columnas tipo str → category ---")
for col, validos in valores_validos_str.items():
    mask = ~df_heart[col].isin(validos) & df_heart[col].notna()
    n = mask.sum()
    print(f"  {'✅' if n == 0 else '❌'} {col:12}: {n:4} filas inválidas eliminadas")
    if n > 0:
        print(f"     Inválidos encontrados: {df_heart.loc[mask, col].unique().tolist()}")
    df_heart = df_heart[~mask].reset_index(drop=True)
    # Convertir a category después de limpiar
    df_heart[col] = df_heart[col].astype("category")
    print(f"     Categorías válidas: {df_heart[col].cat.categories.tolist()}")

# ------------------------------------------------------------
# 3. SLOPE: str → numérico → category ordinal
#    Valores válidos: 1, 2, 3
# ------------------------------------------------------------

print("\n--- Columna slope (str → category ordinal) ---")
df_heart["slope"] = pd.to_numeric(df_heart["slope"], errors="coerce")
mask_slope = ~df_heart["slope"].isin([1.0, 2.0, 3.0])
n_slope = mask_slope.sum()
print(f"  {'✅' if n_slope == 0 else '❌'} slope       : {n_slope:4} filas inválidas eliminadas")
df_heart = df_heart[~mask_slope].reset_index(drop=True)
# Convertir a category ordinal
df_heart["slope"] = pd.Categorical(df_heart["slope"], categories=[1.0, 2.0, 3.0], ordered=True)
print(f"     Categorías: {df_heart['slope'].cat.categories.tolist()} (ordinal)")

# ------------------------------------------------------------
# 4. EXANG: str → bool (0 = No, 1 = Yes)
# ------------------------------------------------------------

print("\n--- Columna exang (str → boolean) ---")
df_heart["exang"] = pd.to_numeric(df_heart["exang"], errors="coerce")
mask_exang = ~df_heart["exang"].isin([0.0, 1.0]) & df_heart["exang"].notna()
n_exang = mask_exang.sum()
print(f"  {'✅' if n_exang == 0 else '❌'} exang       : {n_exang:4} filas inválidas eliminadas")
df_heart = df_heart[~mask_exang].reset_index(drop=True)
df_heart["exang"] = df_heart["exang"].astype("boolean")

# ------------------------------------------------------------
# 5. CA: str → Int8 nullable (valores válidos: 0, 1, 2, 3)
# ------------------------------------------------------------

print("\n--- Columna ca (str → Int8 nullable) ---")
df_heart["ca"] = pd.to_numeric(df_heart["ca"], errors="coerce")
mask_ca = ~df_heart["ca"].isin([0.0, 1.0, 2.0, 3.0]) & df_heart["ca"].notna()
n_ca = mask_ca.sum()
print(f"  {'✅' if n_ca == 0 else '❌'} ca          : {n_ca:4} filas inválidas eliminadas")
df_heart = df_heart[~mask_ca].reset_index(drop=True)
df_heart["ca"] = df_heart["ca"].astype("Int8")

# ============================================================
# RESUMEN FINAL
# ============================================================

print(f"\n{'=' * 60}")
print("RESUMEN FINAL")
print("=" * 60)
print(f"✅ Filas restantes: {len(df_heart)}")
print("\n📋 Tipos de datos finales:")
print(df_heart.dtypes)
print("\n📊 Valores nulos por columna:")
print(df_heart.isnull().sum())
```

Unifica la representación de valores nulos, convirtiendo tanto los nulos de Pandas como los valores '?' a NaN estándar.
Convierte las columnas a sus tipos de datos adecuados: numéricas a float, categóricas a category, y la columna objetivo a bool.
Explora y corrige problemas en los datos: elimina filas duplicadas y maneja valores atípicos extremos usando el método del rango intercuartil.
